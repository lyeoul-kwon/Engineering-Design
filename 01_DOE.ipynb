# D-optimal

```python
import numpy as np
from itertools import product, combinations, permutations #시간 효율성 매우 낮음
import math
```

```python
num_factors = 3  #설계 인자의 개수
num_samples = 15  #선택할 실험 점의 개수
levels = [-1, 0, 1]  #각 요인의 수준 설정
```

```python
def generate_design_space(num_factors, levels):
    design_space = np.array(list(product(levels, repeat=num_factors)))
    return design_space
```

```python
def calculate_d_optimal_score(selected_samples):
    X = np.hstack((np.ones((selected_samples.shape[0], 1)), selected_samples))
    XT_X = np.dot(X.T, X)
    det_XT_X = np.linalg.det(XT_X)
    return -det_XT_X
```

```python
design_space = generate_design_space(num_factors, levels)
```

```python
def find_d_optimal_design(design_space, num_samples):
    best_score = np.inf
    best_design = None

    all_combinations = list(combinations(design_space, num_samples))

    num_combinations_to_check = math.ceil(len(all_combinations) / num_factors)
    selected_combinations = all_combinations[:num_combinations_to_check]

    for selected_samples in selected_combinations:
        selected_samples = np.array(selected_samples)
        score = calculate_d_optimal_score(selected_samples)

        if score < best_score:
            best_score = score
            best_design = selected_samples

    return best_design, -best_score
```

```python
# 설계 공간 생성
design_space = generate_design_space(num_factors, levels)

# D-최적 설계 찾기
best_design, best_score = find_d_optimal_design(design_space, num_samples)

print("Best D-optimal design:")
print(best_design)
print("Best D-optimal score:", best_score)
```
